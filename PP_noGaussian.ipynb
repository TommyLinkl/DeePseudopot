```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt 
%matplotlib inline

qSpace_pot = np.loadtxt("CALCS/CsPbI3_ultraSmall_round2/mc1_heat5_anneal4_results/best_qSpace_pot.dat")
q_coord = qSpace_pot[:,0]
pot_Cs = qSpace_pot[:,1]
pot_I = qSpace_pot[:,2]
pot_Pb = qSpace_pot[:,3]

fig, axs = plt.subplots(1,2, figsize=(10,4))
axs[0].plot(q_coord, pot_Cs, label="Cs, with Gaussian")
axs[0].plot(q_coord, pot_I, label="I, with Gaussian")
axs[0].plot(q_coord, pot_Pb, label="Pb, with Gaussian")
axs[0].set(xlabel=r"$q$", ylabel=r"$v(q)$", title="With Gaussian")
axs[0].grid(alpha=0.5)
axs[0].legend()


noGauss_qSpace_pot = np.loadtxt("CALCS/CsPbI3_ultraSmall_round2/mc1_heat5_anneal4_results/qSpace_pot_noGaussian.dat")
noGauss_q_coord = noGauss_qSpace_pot[:,0]
noGauss_pot_Cs = noGauss_qSpace_pot[:,1]
noGauss_pot_I = noGauss_qSpace_pot[:,2]
noGauss_pot_Pb = noGauss_qSpace_pot[:,3]

axs[1].plot(noGauss_q_coord, noGauss_pot_Cs)
axs[1].plot(noGauss_q_coord, noGauss_pot_I)
axs[1].plot(noGauss_q_coord, noGauss_pot_Pb)
axs[1].set(xlabel=r"$q$", ylabel=r"$v(q)$")
axs[1].grid(alpha=0.5)


gaussian_std = 2.0
divide_pot_Cs = pot_Cs / (np.exp(-q_coord**2 / (2*gaussian_std**2)))
divide_pot_I = pot_I / (np.exp(-q_coord**2 / (2*gaussian_std**2)))
divide_pot_Pb = pot_Pb / (np.exp(-q_coord**2 / (2*gaussian_std**2)))
axs[1].plot(q_coord[:1095], divide_pot_Cs[:1095], ":", label="Cs, divide by Gauss")
axs[1].plot(q_coord[:1095], divide_pot_I[:1095], ":", label="I, divide by Gauss")
axs[1].plot(q_coord[:1095], divide_pot_Pb[:1095], ":", label="Pb, divide by Gauss")
axs[1].set(xlabel=r"$q$", ylabel=r"$v(q)$", title="Without Gaussian")
axs[1].grid(alpha=0.5)
axs[1].legend()

fig.tight_layout()
```
